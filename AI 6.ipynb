To implement the Depth First Search Traversal algorithm in A with suitable example

```python
# Graph representation
graph = {
    '5': ['3', '7'],
    '3': ['2', '4'],
    '7': ['8'],
    '2': [],
    '4': ['8'],
    '8': []
}

# Set to keep track of visited nodes
visited = set()

# Depth First Search (DFS) function
def dfs(visited, graph, node):
    if node not in visited:
        print(node, end=" ")  # Print the node
        visited.add(node)  # Mark the node as visited
        # Recur for all the adjacent nodes (neighbors)
        for neighbour in graph[node]:
            dfs(visited, graph, neighbour)

# Driver code to perform DFS starting from node '5'
print("Following is the Depth First Search traversal:")
dfs(visited, graph, '5')
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F', 'G'],
    'D': ['B'],
    'E': ['B'],
    'F': ['C'],
    'G': ['C']
}
# DFS function (recursive approach)
def dfs(graph, node, visited=None):
    if visited is None:
        visited = set()  # Initialize visited set if it's not passed
    
    # Mark the current node as visited
    visited.add(node)
    print(node, end=" ")  # Print the node
    
    # Recur for all the neighbors of the current node
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

# Driver code to perform DFS starting from node 'A'
print("DFS Traversal starting from node A:")
dfs(graph, 'A')
```
