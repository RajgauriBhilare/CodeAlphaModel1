```python
import pandas as pd

# Load the dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset  (1).csv")

# Split the "Cuisines" column into individual cuisine types
cuisine_series = data['Cuisines'].dropna().str.split(', ').explode()

# Count occurrences of each cuisine
cuisine_counts = cuisine_series.value_counts()

# Get the total number of restaurants
total_restaurants = len(data)

# Get the top 3 cuisines and their percentages
top_cuisines = cuisine_counts.head(3)
top_cuisines_percentage = (top_cuisines / total_restaurants) * 100

# Combine results into a DataFrame
top_cuisines_df = pd.DataFrame({'Cuisine': top_cuisines.index, 
                                'Count': top_cuisines.values, 
                                'Percentage': top_cuisines_percentage.values})

print(top_cuisines_df)
```

```python
import pandas as pd

# Load the dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset  (1).csv")

# Identify the city with the highest number of restaurants
city_counts = data['City'].value_counts()
top_city = city_counts.idxmax()  # City with the most restaurants

# Calculate the average rating for restaurants in each city
city_avg_rating = data.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False)

# Identify the city with the highest average rating
top_rated_city = city_avg_rating.idxmax()  # City with the highest average rating

# Create a DataFrame with results
city_analysis_df = pd.DataFrame({
    'City': city_avg_rating.index,
    'Average Rating': city_avg_rating.values,
    'Total Restaurants': city_counts.reindex(city_avg_rating.index, fill_value=0).values
})

print(f"City with the highest number of restaurants: {top_city}")
print(f"City with the highest average rating: {top_rated_city}")
print("\nTop 5 Cities by Average Rating:\n", city_avg_rating.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset  (1).csv")


# Count the number of restaurants in each price range category
price_range_counts = data['Price range'].value_counts().sort_index()

# Calculate the percentage of restaurants in each price range category
price_range_percentage = (price_range_counts / len(data)) * 100

# Plot the distribution as a bar chart
plt.figure(figsize=(8, 5))
plt.bar(price_range_counts.index, price_range_counts.values, color=['blue', 'green', 'orange', 'red'])
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Price Ranges Among Restaurants")
plt.xticks(price_range_counts.index)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Create a DataFrame for reference
price_range_df = pd.DataFrame({
    'Price Range': price_range_counts.index,
    'Count': price_range_counts.values,
    'Percentage': price_range_percentage.values
})

print(price_range_df)
```

```python
import pandas as pd

# Load the dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset  (1).csv")
# Count the number of restaurants that offer online delivery
online_delivery_counts = data['Has Online delivery'].value_counts()

# Calculate the percentage of restaurants offering online delivery
online_delivery_percentage = (online_delivery_counts / len(data)) * 100

# Compare the average ratings of restaurants with and without online delivery
avg_rating_online = data[data['Has Online delivery'] == 'Yes']['Aggregate rating'].mean()
avg_rating_no_online = data[data['Has Online delivery'] == 'No']['Aggregate rating'].mean()

# Store the results in a dictionary
online_delivery_analysis = {
    "Percentage with Online Delivery": online_delivery_percentage.get('Yes', 0),
    "Percentage without Online Delivery": online_delivery_percentage.get('No', 0),
    "Average Rating (With Online Delivery)": avg_rating_online,
    "Average Rating (Without Online Delivery)": avg_rating_no_online
}

print(online_delivery_analysis)
```
