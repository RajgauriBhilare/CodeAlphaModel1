```python
import pandas as pd

# Example dataset (replace with your actual data loading code)
# data = pd.read_csv('your_dataset.csv')
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset .csv")

# 1. Handle missing values in 'Cuisines' and 'Aggregate rating'
data['Cuisines'] = data['Cuisines'].fillna('')  # Replace NaN with an empty string
data['Aggregate rating'] = pd.to_numeric(data['Aggregate rating'], errors='coerce')  # Convert to numeric
data['Aggregate rating'] = data['Aggregate rating'].fillna(0)  # Handle NaN ratings

# 2. Clean and normalize the 'Cuisines' column
def clean_cuisines(cuisines):
    # Ensure input is a string, clean unwanted characters, and split into sorted tuples
    if isinstance(cuisines, str):
        cuisines = cuisines.strip()  # Remove leading/trailing spaces
        cuisines = cuisines.replace('"', '').replace("'", "")  # Remove quotes
        cuisines_list = [c.strip() for c in cuisines.split(',')]  # Split and trim each item
        return tuple(sorted(filter(None, cuisines_list)))  # Sort and remove empty strings
    return ()

data['Cuisines'] = data['Cuisines'].apply(clean_cuisines)

# 3. Verify the cleaned 'Cuisines' column
print("Sample cleaned 'Cuisines' column:")
print(data['Cuisines'].head(10))

# 4. Group by cuisine combinations and calculate the average rating
grouped_data = data.groupby('Cuisines')['Aggregate rating'].mean().reset_index()

# 5. Sort by average rating
sorted_grouped_data = grouped_data.sort_values(by='Aggregate rating', ascending=False)

# 6. Display the top cuisine combinations
print("Cuisine combinations with the highest average ratings:")
print(sorted_grouped_data.head(10))
```

```python
import pandas as pd
import folium

# 2. Handle missing or invalid latitude and longitude
data['Longitude'] = pd.to_numeric(data['Longitude'], errors='coerce')
data['Latitude'] = pd.to_numeric(data['Latitude'], errors='coerce')
data = data.dropna(subset=['Longitude', 'Latitude'])

# 3. Create a map centered at the average location
average_lat = data['Latitude'].mean()
average_long = data['Longitude'].mean()
restaurant_map = folium.Map(location=[average_lat, average_long], zoom_start=12)

# 4. Define color categories based on ratings
def get_marker_color(rating):
    if rating >= 4.0:
        return 'green'  # High rating
    elif rating >= 3.0:
        return 'orange'  # Medium rating
    else:
        return 'red'  # Low rating

# 5. Add restaurant locations to the map
for _, row in data.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=folium.Popup(
            f"""
            <b>Restaurant:</b> {row['Restaurant Name']}<br>
            <b>Rating:</b> {row['Aggregate rating']}<br>
            <b>Cuisines:</b> {row.get('Cuisines', 'N/A')}<br>
            <b>Cost for Two:</b> ₹{row.get('Average Cost for two', 'N/A')}
            """, max_width=300
        ),
        tooltip=row['Restaurant Name'],
        icon=folium.Icon(color=get_marker_color(row['Aggregate rating']), icon='info-sign')
    ).add_to(restaurant_map)

# 6. Save and display the map
restaurant_map.save("restaurant_locations_map.html")
restaurant_map
```
