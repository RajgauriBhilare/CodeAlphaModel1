```python
#13.number is pronic or not 
# Input number from user
num = int(input("Enter a number: "))

# Check if the number is pronic
n = 0
while n * (n + 1) < num:
    n += 1

if n * (n + 1) == num:
    print(f"{num} is pronic number")
else:
    print(f"{num} is not pronic number")
 

```

```python
#17.Python Program to Form the Factorial Series:
# Input: Get the number 'n' from the user
n = int(input("Enter a number to generate the factorial series: "))

# Initialize a variable to calculate the factorial
factorial = 1

# Generate and display the factorial series
for i in range(1, n + 1):
    factorial *= i  # Multiply the current factorial value by i
    print(f"Factorial of {i} is {factorial}")
```

```python
#18.Python program to display the Fibonacci sequence up to the n-th term:
# Input: Get the number 'n' from the user
n = int(input("Enter the number of terms in the Fibonacci sequence: "))

# Initialize the first two terms
a, b = 0, 1

# Display the Fibonacci sequence up to the n-th term
print("Fibonacci sequence:")
for i in range(n):
    print(a, end=" ")
    # Update the values of a and b
    a, b = b, a + b
```
