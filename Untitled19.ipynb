```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a small dataset
data = {
    'Feature A': [20, 30, 25, 15],
    'Feature B': [30, 25, 35, 40],
    'Feature C': [25, 20, 15, 25],
    'Feature D': [25, 25, 25, 20]
}

# Create a DataFrame
df_grp = pd.DataFrame(data, index=['Group 1', 'Group 2', 'Group 3', 'Group 4'])

# Define colors and explode for pie chart
colors = ['gold', 'lightcoral', 'yellowgreen', 'lightskyblue']
explode = (0, 0, 0, 0.1)  # only "explode" the 4th slice

# Create subplots
fig, [[ax1, ax2], [ax3, ax4]] = plt.subplots(nrows=2, ncols=2)
ax = [ax1, ax2, ax3, ax4]

# Titles for each pie chart
type_1_names = df_grp.index.tolist()

# Loop through each group to create pie charts
for i in range(4):
    percent = df_grp.iloc[i, :]
    ax[i].pie(percent, explode=explode,
               labels=percent.index, colors=colors,
               autopct='%.2f%%',  # display value
               shadow=True,
               startangle=90)
    ax[i].set_aspect('equal')
    ax[i].set_title(type_1_names[i])

# Super title for the figure
plt.suptitle('Comparing Major Features of Pokémon Groups',
             fontsize=14,
             fontweight='bold')

# Show the plot
plt.tight_layout(rect=[0, 0.03, 1, 0.95])  # Adjust layout
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model

# Create a simple synthetic dataset
# Features: Sepal Width and Sepal Height (just two points for simplicity)
X = np.array([[1, 2], [2, 3], [3, 1], [5, 4], [6, 5], [7, 3]])
Y = np.array([0, 0, 0, 1, 1, 1])  # Two classes: 0 and 1

# Create and fit the logistic regression model
lm = linear_model.LogisticRegression(C=1e5)
lm.fit(X, Y)

# Range of X and Y axis
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

# Step size in the mesh
h = 0.1  

# Creating Mesh Grid
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))

# Predict the class for each point in the mesh grid
Z = lm.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plotting
plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)  # Decision boundary
plt.scatter(X[:, 0], X[:, 1], c=Y, edgecolors='k', marker='o', cmap=plt.cm.coolwarm)  # Data points
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Logistic Regression Decision Boundary with Simple Data')
plt.show()
```
