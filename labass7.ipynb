```python
#2.Conduct a univariate analysis of both numerical and categorical variables. How are the numerical
#variables distributed, and what is the frequency distribution of the categorical variables?

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data=pd.read_csv("C:/Users/Rajguru/Downloads/cards_data.csv")
data

# Step 1: Identify numerical columns
numerical_cols = data.select_dtypes(include=['int64', 'float64']).columns

# Step 2: Summary statistics of numerical variables
print("Summary statistics for numerical variables:")
print(data[numerical_cols].describe())
    
```

```python
# Step 3: Distribution of numerical variables
for col in numerical_cols:
    plt.figure(figsize=(12, 6))
    
# Histogram
plt.subplot(1, 2, 1)
sns.histplot(data[col], kde=True, bins=20, color='skyblue')
plt.title(f'Distribution of {col}')
    
# Box plot
plt.subplot(1, 2, 2)
sns.boxplot(data[col], color='lightgreen')
plt.title(f'Boxplot of {col}')
    
plt.tight_layout()
plt.show()
```

```python
# Step 1: Identify categorical columns
categorical_cols = data.select_dtypes(include=['object']).columns

# Step 2: Frequency distribution of categorical variables
for col in categorical_cols:
    print(f"Frequency distribution for {col}:")
    print(data[col].value_counts())  # Frequency of each category
    
# Bar plot
plt.figure(figsize=(8, 6))
sns.countplot(data=data, x=col, palette="Set2")
plt.title(f'Frequency Distribution of {col}')
plt.xticks(rotation=45)
plt.show()
```

```python
#QUESTION 3.3 Analyze the relationships between the numerical variables. What correlations exist between these
#variables, and how can these relationships be visualized using scatter plots and a correlation matrix?

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
data=pd.read_csv("C:/Users/Rajguru/Downloads/cards_data.csv")
data
# Step 1: Identify numerical columns
numerical_cols = data.select_dtypes(include=['int64', 'float64']).columns
# Step 2: Calculate the correlation matrix
correlation_matrix = data[numerical_cols].corr()
# Step 3: Plot the correlation matrix as a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix")
plt.show()
# Step 4: Visualize relationships using scatter plots
# For all pairs of numerical variables, we'll create scatter plots
for i in range(len(numerical_cols)):
    for j in range(i + 1, len(numerical_cols)):
        plt.figure(figsize=(8, 6))
sns.scatterplot(x=data[numerical_cols[i]], y=data[numerical_cols[j]], color="blue")
plt.title(f"Scatter Plot: {numerical_cols[i]} vs {numerical_cols[j]}")
plt.xlabel(numerical_cols[i])
plt.ylabel(numerical_cols[j])
plt.show()
```

```python
#question 4.Identify outliers in the dataset. How will you detect outliers in the numerical variables using visual or
#statistical methods, and what steps will you take to handle these outliers?

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
data=pd.read_csv("C:/Users/Rajguru/Downloads/cards_data.csv")
data
```

```python
# Step 1: Identify numerical columns
numerical_cols = data.select_dtypes(include=['int64', 'float64']).columns

# Step 2: Plot boxplots for numerical variables
for col in numerical_cols:
    plt.figure(figsize=(8, 6))
sns.boxplot(x=data[col], color='lightgreen')
plt.title(f'Boxplot for {col}')
plt.show()
```

```python
from scipy.stats import zscore

# Step 1: Calculate the Z-scores for numerical columns
z_scores = data[numerical_cols].apply(zscore)

# Step 2: Identify outliers (Z-score > 3 or < -3)
outliers_zscore = (z_scores > 3) | (z_scores < -3)

# Step 3: Show the rows with outliers
outlier_rows = data[outliers_zscore.any(axis=1)]
print(f"Rows with outliers based on Z-score:\n{outlier_rows}")
```

```python
# Step 1: Calculate IQR for each numerical column
Q1 = data[numerical_cols].quantile(0.25)
Q3 = data[numerical_cols].quantile(0.75)
IQR = Q3 - Q1

# Step 2: Identify outliers
outliers_iqr = ((data[numerical_cols] < (Q1 - 1.5 * IQR)) | (data[numerical_cols] > (Q3 + 1.5 * IQR)))

# Step 3: Show rows with outliers
outlier_rows_iqr = data[outliers_iqr.any(axis=1)]
print(f"Rows with outliers based on IQR:\n{outlier_rows_iqr}")
```

```python
# Replot boxplots to visualize the effect of handling outliers
for col in numerical_cols:
    plt.figure(figsize=(8, 6))
sns.boxplot(x=data[col], color='lightgreen')
plt.title(f'Boxplot for {col} After Handling Outliers')
plt.show()
```
