```python
import matplotlib.pyplot as plt
import networkx as nx

def create_lifecycle_diagram():
    # Define the steps in the Data Science Life Cycle
    steps = [
        "Problem Identification",
        "Data Collection",
        "Data Cleaning & Preprocessing",
        "Exploratory Data Analysis (EDA)",
        "Feature Engineering",
        "Model Building",
        "Model Evaluation",
        "Deployment"
    ]
    
    # Create a directed graph
    G = nx.DiGraph()
    
    # Add nodes and edges
    for i in range(len(steps) - 1):
        G.add_edge(steps[i], steps[i + 1])
    
    # Set up the figure
    plt.figure(figsize=(10, 6))
    
    # Generate layout
    pos = nx.spring_layout(G, seed=42)
    
    # Draw the nodes and edges
    nx.draw(G, pos, with_labels=True, node_size=3000, node_color="lightblue", edge_color="gray", font_size=10, font_weight='bold')
    
    # Set title
    plt.title("Data Science Life Cycle", fontsize=14, fontweight='bold')
    
    # Show the diagram
    plt.show()

# Run the function to create the diagram
create_lifecycle_diagram()
```

```python
import matplotlib.pyplot as plt
import networkx as nx

def create_lifecycle_diagram():
    # Define the steps in the Data Science Life Cycle
    steps = [
        "Problem Identification",
        "Data Collection",
        "Data Cleaning & Preprocessing",
        "Exploratory Data Analysis (EDA)",
        "Feature Engineering",
        "Model Building",
        "Model Evaluation",
        "Deployment"
    ]
    
    # Create a directed graph
    G = nx.DiGraph()
    
    # Add nodes and edges in a circular manner
    for i in range(len(steps)):
        G.add_edge(steps[i], steps[(i + 1) % len(steps)])  # Connect last node to first
    
    # Set up the figure
    plt.figure(figsize=(8, 8))
    
    # Generate circular layout
    pos = nx.circular_layout(G)
    
    # Draw the nodes and edges
    nx.draw(G, pos, with_labels=True, node_size=3000, node_color="lightblue", edge_color="gray", font_size=10, font_weight='bold')
    
    # Set title
    plt.title("Data Science Life Cycle", fontsize=14, fontweight='bold')
    
    # Show the diagram
    plt.show()

# Run the function to create the diagram
create_lifecycle_diagram()
```
