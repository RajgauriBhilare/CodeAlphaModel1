```python
from collections import deque

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = set([start])
    

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(path + [neighbor])

    return None

# Example usage:
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

start = 'A'
goal = 'G'

shortest_path = bfs(graph, start, goal)
print(shortest_path)  # Output: ['A', 'B', 'C', 'F']

```
