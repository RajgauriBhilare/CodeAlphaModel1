Assignment 11

Solve 8 -puzzle problem using Breadth First Search

```python
from collections import deque

# Goal state for the 8-puzzle
goal_state = [1, 2, 3, 4, 5, 6, 7, 8,0]

# Directions for moving the empty space (0): up, down, left, right
moves = [(-3, 'up'), (3, 'down'), (-1, 'left'), (1, 'right')]

# Check if a state is the goal state
def is_goal(state):
    return state == goal_state

# Get all valid neighbors (states) by moving the empty space
def get_neighbors(state):
    neighbors = []
    index = state.index(0)  # Find the empty space (0)
    
    for move, direction in moves:
        new_index = index + move
        
        # Ensure the move is valid (not out of bounds or across rows)
        if (move == -1 and index % 3 == 0) or (move == 1 and index % 3 == 2):
            continue
        
        if 0 <= new_index < len(state):
            # Swap the empty space with the adjacent tile
            new_state = state[:]
            new_state[index], new_state[new_index] = new_state[new_index], new_state[index]
            neighbors.append(new_state)
    
    return neighbors

# BFS to solve the puzzle
def bfs(initial_state):
    queue = deque([(initial_state, [])])  # (state, path to reach the state)
    visited = set()
    visited.add(tuple(initial_state))  # Store visited states as tuples

    while queue:
        current_state, path = queue.popleft()

        # If we reached the goal state, return the solution path
        if is_goal(current_state):
            return path + [current_state]

        # Explore neighbors (valid moves)
        for neighbor in get_neighbors(current_state):
            if tuple(neighbor) not in visited:
                visited.add(tuple(neighbor))
                queue.append((neighbor, path + [current_state]))
    
    return None  # Return None if no solution is found

# Function to print the state in 3x3 grid format
def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()

# Example initial state (unsolved puzzle)
initial_state = [1, 2, 3, 4, 5, 6, 0, 7, 8]

# Print the initial state
print("Initial State:")
print_puzzle(initial_state)

# Solve the puzzle using BFS
solution = bfs(initial_state)

if solution:
    print("Solution found:")
    for step in solution:
        print_puzzle(step)
else:
    print("No solution found.")
```
