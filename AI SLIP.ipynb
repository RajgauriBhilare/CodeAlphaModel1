```python
# Write a python program to implement Water Jug Problem.
def water_jug_problem(jug1_capacity, jug2_capacity, target):
    # Initialize the state with both jugs empty
    jug1, jug2 = 0, 0
    steps = []

    while jug1 != target and jug2 != target:
        steps.append(f"Jug1: {jug1}, Jug2: {jug2}")

        # If Jug1 is empty, fill it
        if jug1 == 0:
            jug1 = jug1_capacity
        # If Jug2 is full, empty it
        elif jug2 == jug2_capacity:
            jug2 = 0
        # If Jug1 has more water than Jug2 can hold, transfer water from Jug1 to Jug2
        elif jug1 > 0 and jug2 < jug2_capacity:
            transfer = min(jug1, jug2_capacity - jug2)
            jug1 -= transfer
            jug2 += transfer

    steps.append(f"Jug1: {jug1}, Jug2: {jug2}")
    return steps

# Example usage
jug1_capacity = 4  # Jug1 capacity (liters)
jug2_capacity = 3  # Jug2 capacity (liters)
target = 2  # Target amount of water to measure

steps = water_jug_problem(jug1_capacity, jug2_capacity, target)
for step in steps:
    print(step)
```

```python
#2. Write a python program to implement a 8 -puzzle problem using Breadth First Search.

class Solution:
    def solve(self, board):
        # Define the target state
        target = (0, 1, 2, 3, 4, 5, 6, 7, 8)
        # Dictionary to store visited states and their distances
        visited = {}
        # Flatten the input board into a tuple
        flatten = tuple([num for row in board for num in row])
        # Mark the starting state as visited
        visited[flatten] = 0
        # If the initial state is already the target, return 0 moves
        if flatten == target:
            return 0
        # Perform BFS to find the shortest path to the target
        return self.get_paths(visited, target)

    def get_paths(self, visited, target):
        cnt = 0
        while True:
            # Find all nodes at the current distance
            current_nodes = [node for node in visited if visited[node] == cnt]
            if not current_nodes:
                return -1  # If no nodes to process, return failure
            for node in current_nodes:
                # Get all possible next states from the current node
                next_moves = self.find_next(node)
                for move in next_moves:
                    if move not in visited:
                        visited[move] = cnt + 1
                        if move == target:
                            return cnt + 1
            cnt += 1

    def find_next(self, node):
        # Possible moves for each index
        moves = {
            0: [1, 3],
            1: [0, 2, 4],
            2: [1, 5],
            3: [0, 4, 6],
            4: [1, 3, 5, 7],
            5: [2, 4, 8],
            6: [3, 7],
            7: [4, 6, 8],
            8: [5, 7],
        }
        results = []
        pos_0 = node.index(0)  # Find the position of the empty tile (0)
        for move in moves[pos_0]:
            # Swap the empty tile with the target position
            new_node = list(node)
            new_node[pos_0], new_node[move] = new_node[move], new_node[pos_0]
            results.append(tuple(new_node))
        return results


# Example usage
ob = Solution()
matrix = [[3, 1, 2], [4, 7, 5], [6, 8, 0]]
print(ob.solve(matrix))
```
