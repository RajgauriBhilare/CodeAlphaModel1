```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/MBA.csv")

# Basic dataset information
print("Dataset Shape:", data.shape)
print("Dataset Columns:", data.columns)

# Check for missing values
print("\n--- Missing Values ---")
print(data.isnull().sum())  # Count missing values for each column

# Detect outliers using the IQR method
def detect_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return lower_bound, upper_bound

# Select the column for outlier detection
column = 'race'  # Replace 'YourColumnName' with the column you want to analyze
lower_bound, upper_bound = detect_outliers(data, column)

# Plot box plot to visualize outliers
plt.figure(figsize=(8, 6))
sns.boxplot(data[race], color='skyblue')
plt.axhline(lower_bound, color='red', linestyle='--', label='Lower Bound')
plt.axhline(upper_bound, color='blue', linestyle='--', label='Upper Bound')

# Add labels and title
plt.title(f"Box Plot for {column} with Outlier Thresholds")
plt.xlabel(race)
plt.legend()
plt.show()
```

```python
import pandas  as pd
data={
    'Name':['hhh','kkkk','ggg','uuu','ttt'],
    'age':[21,23,25,22,24],
    'gender':['male','female','female','male','male'],
    'department':['it','hr','hr','finace','it'],
    'salary':[52000,56000,65000,63000,56000],
     }
df=pd.DataFrame(data)
print(df)

#filterd
It_department =df[df['department']=='it']
print("\nfilterd data(It department):",It_department)

#grouping
grouped_data=df.groupby('department')['salary'].mean()
print("grouped data is:\n",grouped_data)

#aggregating
aggregated_data=df.groupby('gender')['salary'].agg(['mean','max'])
print("aggregated data is the:\n",aggregated_data)
```

```python
import pandas as pd
import numpy as np

# Example dataset with missing values
data = {
    'Age': [25, np.nan, 30, 35, np.nan],
    'Salary': [50000, 60000, np.nan, 80000, 90000],
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Detect missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Impute missing values
df['Age'] = df['Age'].fillna(df['Age'].mean())  # Impute Age with mean
df['Salary'] = df['Salary'].fillna(df['Salary'].median())  # Impute Salary with median

print("\nDataFrame after Imputation:")
print(df)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns  # Import seaborn
from scipy.stats import skew

# Create a skewed dataset
df = pd.DataFrame({'Feature': np.random.exponential(scale=2, size=1000)})

# Plot a histogram with KDE
plt.figure(figsize=(10, 6))  # Corrected figsize argument
sns.histplot(df['Feature'], kde=True, color='skyblue', bins=30)
plt.title("Histogram and KDE Plot of Skewed Data")
plt.xlabel("Feature")
plt.ylabel("Frequency")
plt.show()

# Plot a box plot
plt.figure(figsize=(10, 6))  # Corrected figsize argument
sns.boxplot(x=df['Feature'], color='pink')
plt.title("Box Plot of Skewed Data")
plt.xlabel("Feature")
plt.show()
```
