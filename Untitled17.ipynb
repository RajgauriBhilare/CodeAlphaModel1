```python
import pandas as pd
import matplotlib.pyplot as plt
# Load your dataset (replace 'your_dataset.csv' with your actual file path) 
data = pd.read_csv("C:/Users/Rajguru/Downloads/MBA.csv")
# Group the data by gender and count the occurrences
gender_counts = data['gender'].value_counts()
# Create the bar chart
plt.bar(gender_counts.index, gender_counts.values)
# Customize the chart
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
# Show the chart
plt.show()
```

```python
#Pie
import pandas as pd
import matplotlib.pyplot as plt
# Load the dataset (replace 'your_dataset.csv' with your actual file path)
data = pd.read_csv("C:/Users/Rajguru/Downloads/MBA.csv")
# Group the data by gender and count the occurrences
gender_counts = data['gender'].value_counts()
# Create the pie chart
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%')
# Set the title
plt.title('Gender Distribution')
# Show the chart
plt.show()
```

```python
#Line 
import pandas as pd
import matplotlib.pyplot as plt
# Load the dataset (replace 'your_dataset.csv' with your actual file path)
data =pd.read_csv("C:/Users/Rajguru/Downloads/MBA.csv")
# Group the data by gender and calculate the mean GPA
grouped_data = data.groupby('gender')['gpa'].mean()
# Create the line chart
plt.plot(grouped_data.index, grouped_data.values, marker='o', linestyle='-', color='blue')
# Customize the chart
plt.title('Average GPA by Gender')
plt.xlabel('Gender')
plt.ylabel('Average GPA')
# Show the chart
plt.show()
```
