```python
from collections import deque

def bidirectional_search(graph, start, goal):
    if start == goal:
        return [start]

    # Initialize forward and backward queues
    forward_queue = deque([start])
    backward_queue = deque([goal])

    # Visited sets to track visited nodes
    visited_forward = {start: None}
    visited_backward = {goal: None}

    while forward_queue and backward_queue:
        # Explore forward from the start
        if forward_queue:
            current_forward = forward_queue.popleft()
            for neighbor in graph[current_forward]:
                if neighbor not in visited_forward:
                    visited_forward[neighbor] = current_forward
                    forward_queue.append(neighbor)

                    # Check if this neighbor has been visited by backward search
                    if neighbor in visited_backward:
                        return reconstruct_path(visited_forward, visited_backward, neighbor)

        # Explore backward from the goal
        if backward_queue:
            current_backward = backward_queue.popleft()
            for neighbor in graph[current_backward]:
                if neighbor not in visited_backward:
                    visited_backward[neighbor] = current_backward
                    backward_queue.append(neighbor)

                    # Check if this neighbor has been visited by forward search
                    if neighbor in visited_forward:
                        return reconstruct_path(visited_forward, visited_backward, neighbor)

    return None  # No path found


def reconstruct_path(visited_forward, visited_backward, meeting_node):
    # Build the path from start to meeting_node
    path = []
    # Forward path
    current = meeting_node
    while current is not None:
        path.append(current)
        current = visited_forward[current]
    path.reverse()  # Reverse to get path from start to meeting_node

    # Backward path
    current = visited_backward[meeting_node]
    while current is not None:
        path.append(current)
        current = visited_backward[current]

    return path


# Example graph as an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'G', 'H'],
    'F': ['C'],
    'G': ['E'],
    'H': ['E']
}

# Test the bidirectional search
start_node = 'A'
goal_node = 'F'
path = bidirectional_search(graph, start_node, goal_node)

if path:
    print("Shortest path from", start_node, "to", goal_node, ":", path)
else:
    print("No path found from", start_node, "to", goal_node)
```
