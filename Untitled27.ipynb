```python
import matplotlib.pyplot as plt

# Data
X = [68, 64, 75, 50, 64, 80, 75, 40, 55, 64]
Y = [62, 58, 68, 45, 81, 60, 68, 48, 50, 70]

# Create scatter plot
plt.scatter(X, Y, color='blue', marker='o')

# Add titles and labels
plt.title('Scatter Diagram of X vs Y')
plt.xlabel('X Values')
plt.ylabel('Y Values')

# Show grid
plt.grid(True)

# Show the plot
plt.show()
```
