```python
from collections import deque

# BFS function
def bfs(graph, start):
    # Create a queue for BFS and a set to track visited nodes
    visited = set()  # To track visited nodes
    queue = deque([start])  # Initialize the queue with the start node

    # List to store the BFS traversal result
    bfs_result = []

    while queue:
        # Dequeue the first element
        node = queue.popleft()

        # If the node has not been visited, visit it
        if node not in visited:
            visited.add(node)  # Mark the node as visited
            bfs_result.append(node)  # Add to the result list

            # Enqueue all unvisited neighbors
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)

    return bfs_result

# Example Graph represented as an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F', 'G'],
    'D': ['B'],
    'E': ['B'],
    'F': ['C'],
    'G': ['C']
}

# Start BFS from node 'A'
start_node = 'A'
print(f"BFS Traversal starting from node {start_node}: {bfs(graph, start_node)}")
```
