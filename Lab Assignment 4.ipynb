```python
#1.Represent of statistical raw data of bar chart

import matplotlib.pyplot as plt

# Example of raw data
categories = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
values = [23, 45, 56, 78, 33]

# Create a bar chart
plt.bar(categories, values, color='skyblue')

# Add title and labels
plt.title('Raw Data Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

# Show the bar chart
plt.show()
```

```python
#2.Representation of statistical raw data chart,pie chart etc.


import matplotlib.pyplot as plt

# Example raw data
categories = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
values = [23, 45, 56, 78, 33]

# 1. Bar Chart
plt.figure(figsize=(10, 6))  # Create a figure with specific size
plt.subplot(2, 2, 1)  # Create a subplot in a 2x2 grid, first position
plt.bar(categories, values, color='skyblue')
plt.title('Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

# 2. Pie Chart
plt.subplot(2, 2, 2)  # Second position in the 2x2 grid
plt.pie(values, labels=categories, autopct='%1.1f%%', colors=['lightcoral', 'yellowgreen', 'lightskyblue', 'orange', 'purple'])
plt.title('Pie Chart')

# 3. Line Chart
plt.subplot(2, 2, 3)  # Third position in the 2x2 grid
plt.plot(categories, values, marker='o', linestyle='-', color='b', markersize=8)
plt.title('Line Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

# 4. Horizontal Bar Chart
plt.subplot(2, 2, 4)  # Fourth position in the 2x2 grid
plt.barh(categories, values, color='salmon')
plt.title('Horizontal Bar Chart')
plt.xlabel('Values')
plt.ylabel('Categories')

# Display all charts
plt.tight_layout()  # Adjust spacing between plots
plt.show()
```
