```python
import csv
# Load the dataset
with open("C:/Users/Rajguru/Downloads/MBA.csv", 'r') as csvfile:
 reader = csv.reader(csvfile)
 data = [row for row in reader]
# Specify the column index (assuming column names are in the first row)
column_index = 3 # Replace 1 with the actual index of the 'gpa' column
# Extract values from the specified column
values = [float(row[column_index]) for row in data[1:]]
# Sort the values in ascending order
values.sort()
# Calculate the median based on the number of values
if len(values) % 2 == 0:
 median_value = (values[len(values) // 2 - 1] + values[len(values) // 2]) / 2
else:
 median_value = values[len(values) // 2]
# Print the result
print("The median of the column 'gpa' is:", median_value)
```

```python
import csv
from collections import Counter
# Load the dataset
with open("C:/Users/Rajguru/Downloads/MBA.csv", 'r') as csvfile:
 reader = csv.reader(csvfile)
 data = [row for row in reader]
# Specify the column index (assuming column names are in the first row)
column_index = 6 # Replace 1 with the actual index of the 'gpa' column
# Extract values from the specified column
values = [float(row[column_index]) for row in data[1:]]
# Calculate the mode using the Counter class
mode_value = Counter(values).most_common(1)[0][0]
# Print the result
print("The mode of the column 'gpa' is:", mode_value)
```
