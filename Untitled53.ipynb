```python
import numpy as np

def travellingsalesman(c):
    global cost
    adj_vertex = -1
    min_val = float('inf')
    visited[c] = 1
    print((c + 1), end=" -> ")
    
    for k in range(n):
        if tsp_g[c][k] != 0 and visited[k] == 0:
            if tsp_g[c][k] < min_val:
                min_val = tsp_g[c][k]
                adj_vertex = k

    if adj_vertex != -1:
        cost += min_val
        travellingsalesman(adj_vertex)
    else:
        # Return to the starting point
        cost += tsp_g[c][0]
        print(1, end="")  # Printing the return to the starting city

# Input graph and initialization
n = 4
cost = 0
visited = np.zeros(n, dtype=int)
tsp_g = np.array([
    [0, 20, 40, 22],
    [20, 0, 13, 30],
    [40, 13, 0, 12],
    [22, 30, 12, 0]
])

# Solve the TSP problem
print("Shortest path: ", end="")
travellingsalesman(0)
print()
print("Minimum cost: ", end="")
print(cost)
```

```python
import nltk
from nltk.chat.util import Chat,reflections


chatbot_responses=[
    [r"hello|hii|hey",["Hello","Hi there!","Hey!"]],
    [r"how are you?",["i'm good ,thanks","I'm doing well ,how about you?"]],
    [r"What is your name?",["i'm a chatbot","i dont havr a name","call mi your friend"]],
    [r"bye|goodbye",["GoodBye!","Bye","TakeCare"]]
]
chatbot = Chat(chatbot_responses, reflections)
print("Hello! i am a simple chatbot.You can type'exit' to end conversation.")
while True:
    user_input=input("enter your Message:")
    if user_input.lower()=="exit":
        print("chatbot:Goodbye!")
        break
    response=chatbot.respond(user_input)
    print("Chatbot:",response)
```
