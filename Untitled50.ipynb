```python
#1. Write a Python program to calculate the mean, median, and mode of a dataset using 
#NumPy and Pandas.


import numpy as np
import pandas as pd

# Sample dataset
data = [10, 20, 20, 30, 40, 50, 50, 50, 60]

# Using NumPy
mean_numpy = np.mean(data)
median_numpy = np.median(data)
mode_numpy = max(set(data), key=data.count)  # Simpler way to calculate mode without SciPy

print("Using NumPy:")
print(f"Mean: {mean_numpy}")
print(f"Median: {median_numpy}")
print(f"Mode: {mode_numpy}")

# Using Pandas
series = pd.Series(data)
mean_pandas = series.mean()
median_pandas = series.median()
mode_pandas = series.mode()[0]  # Getting the first mode if multiple exist

print("\nUsing Pandas:")
print(f"Mean: {mean_pandas}")
print(f"Median: {median_pandas}")
print(f"Mode: {mode_pandas}")

```

```python
#1. Write a Python program to differentiate between discrete and continuous attributes in a dataset.

import pandas as pd
import numpy as np

# Create a sample dataset
data = {
    "Age": [25, 30, 22, 40, 35],          # Continuous
    "Income": [50000, 60000, 40000, 80000, 70000],  # Continuous
    "Gender": ["Male", "Female", "Female", "Male", "Female"],  # Discrete (categorical)
    "Number_of_Cars": [1, 2, 0, 1, 3],   # Discrete (numeric)
    "Has_House": [1, 0, 0, 1, 1]         # Discrete (binary indicator)
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Function to classify attributes
def classify_attributes(df):
    discrete = []
    continuous = []
    
    for column in df.columns:
        if df[column].dtypes in ['int64', 'float64']:
            # Check for discrete vs continuous by unique value count
            unique_count = df[column].nunique()
            if unique_count / len(df[column]) < 0.2:  # Arbitrary threshold: fewer unique values = discrete
                discrete.append(column)
            else:
                continuous.append(column)
        else:
            discrete.append(column)  # Non-numeric columns are discrete by default

    return discrete, continuous

# Classify attributes
discrete_attrs, continuous_attrs = classify_attributes(df)

# Display the results
print("Dataset:")
print(df)
print("\nDiscrete Attributes:", discrete_attrs)
print("Continuous Attributes:", continuous_attrs)
```

```python
#. What is normalization, and how does it differ from standardization?
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Sample DataFrame
data = {
    "Feature1": [10, 15, 25, 35],
    "Feature2": [200, 300, 400, 800]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Normalization
min_max_scaler = MinMaxScaler(feature_range=(0, 1))
normalized_df = pd.DataFrame(min_max_scaler.fit_transform(df), columns=df.columns)

print("\nNormalized DataFrame:")
print(normalized_df)

# Standardization
standard_scaler = StandardScaler()
standardized_df = pd.DataFrame(standard_scaler.fit_transform(df), columns=df.columns)

print("\nStandardized DataFrame:")
print(standardized_df)
```
