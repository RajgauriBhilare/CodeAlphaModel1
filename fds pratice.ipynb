```python
import numpy as np
data=[1,2,3]
mean=np.mean(data)
median=np.median(data)
mode=max(set(data),key=data.count)
print(mean)
print(mode)
print(median)
```

```python
import pandas as pd
data=pd.read_csv("C:/Users/Rajguru/Downloads/MBA.csv")
discrete_attributes=[]
continuous_attribute=[]
for column in data.columns:
    if data[column].dtype=='object':
        discrete_attributes.append(column)
    elif data[column].dtype in['int64','float64']:
        unique_values=data[column].nunique()
        if unique_values<=10:
            discrete_attributes.append(column)
        else:
            continuous_attribute.append(column)
print("Discrete attributes:")
print(discrete_attributes)
print(continuous_attribute)
```
