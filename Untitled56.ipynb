```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression

# Load the dataset (replace 'your_data.csv' with the actual file path)
df = pd.read_csv("C:/Users/Rajguru/Downloads/Statitics Mini Project.csv/Statitics Mini Project.csv")

# Data cleaning (check for missing values and handle them)
df.isnull().sum()  # Check for missing values
df = df.dropna()  # Drop rows with missing values (or alternatively, fill them with mean/median)

# Descriptive statistics
descriptive_stats = df.describe()  # Summary statistics (mean, std, min, 25%, 50%, 75%, max)
print(descriptive_stats)

# Demographics Analysis (e.g., gender and age distribution)
sns.countplot(x='gender', data=df)  # Gender distribution plot
plt.title("Gender Distribution")
plt.show()

sns.histplot(df['age'], kde=True)  # Age distribution plot
plt.title("Age Distribution")
plt.show()

# Promotions and Discounts Analysis
promotions_spending = df[df['promotion_spend'] > 0]  # Filter for respondents who spend during promotions
sns.histplot(promotions_spending['promotion_spend'], kde=True, color='green')  # Spending during promotions
plt.title("Spending During Promotions")
plt.show()

# Correlation analysis (e.g., Age and Response to Promotions)
age_promotion_corr, _ = pearsonr(df['age'], df['promotion_spend'])
print(f"Correlation between Age and Promotion Spend: {age_promotion_corr}")

# Visualize Correlation Matrix for multiple variables
corr_matrix = df[['age', 'promotion_spend', 'service_quality', 'loyalty_program']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix")
plt.show()

# Marketing Channel Effectiveness (e.g., influence of social media)
sns.countplot(x='social_media_influence', data=df)  # Marketing Channel Influence (Social Media)
plt.title("Effectiveness of Social Media Marketing")
plt.show()

# Linear Regression Analysis (e.g., Service Quality and Customer Retention)
# Assuming 'service_quality' and 'customer_retention' are columns in the dataset
X = df[['service_quality']]  # Independent variable (Service Quality)
y = df['customer_retention']  # Dependent variable (Customer Retention)

# Linear Regression Model
regressor = LinearRegression()
regressor.fit(X, y)
y_pred = regressor.predict(X)

# Plotting the regression line
plt.scatter(X, y, color='blue', label='Actual data')
plt.plot(X, y_pred, color='red', label='Regression Line')
plt.title('Service Quality vs Customer Retention')
plt.xlabel('Service Quality')
plt.ylabel('Customer Retention')
plt.legend()
plt.show()

# Loyalty Program Influence (e.g., proportion of respondents who find loyalty programs influential)
loyalty_program_influence = df['loyalty_program_influence'].value_counts(normalize=True)
print("Proportion of respondents influenced by loyalty programs:")
print(loyalty_program_influence)

# Recommendations based on analysis (e.g., segmenting by age and promotional responsiveness)
age_groups = pd.cut(df['age'], bins=[18, 25, 35, 45, 60, 100], labels=['18-25', '26-35', '36-45', '46-60', '60+'])
promotion_by_age = df.groupby(age_groups)['promotion_spend'].mean()
print("Average Promotion Spend by Age Group:")
print(promotion_by_age)

# Plotting the promotion spend by age group
promotion_by_age.plot(kind='bar', color='purple')
plt.title("Average Promotion Spend by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Spend During Promotions")
plt.show()
```
