```python
#1.Represent of statistical raw data of bar chart

import matplotlib.pyplot as plt

products = ['A', 'B', 'C', 'D']
sales = [50, 75, 100, 60]

plt.bar(products, sales,color='black')
plt.title('Sales by Product')
plt.xlabel('Products')
plt.ylabel('Number of Sales')
plt.show()

```

```python
#2.Representation of statistical raw data chart,pie chart

import matplotlib.pyplot as plt
products = ['A', 'B', 'C', 'D']
sales = [50, 75, 100, 60]
plt.pie(sales, labels=products)
plt.show()
```

```python
#1.Download CSV File from Kaggle.com and create all types of Chart
import pandas as pd
data = pd.read_csv("C:/Users/Rajguru/Downloads/Spotify Most Streamed Songs.csv")
print(data.head())
```

```python
#Bar chart
plt.figure(figsize=(10, 5))
data['released_year'].value_counts().plot(kind='bar')
plt.title('Bar Chart of Categories')
plt.xlabel('released_year')
plt.ylabel('artist_count')
plt.show()
```

```python
#pie chart
plt.figure(figsize=(4, 4))
plt.pie(data['mode'].value_counts(), labels=data['mode'].value_counts().index, startangle=90)
plt.title('Pie Chart of Categories')
plt.axis('equal')  
plt.show()
```

```python
#scatter plot
import seaborn as sns
plt.figure(figsize=(10, 5))
sns.scatterplot(data=data, x='key', y='in_spotify_playlists')  
plt.title('Scatter Plot of X vs Y')
plt.xlabel('key')
plt.ylabel('in_spotify_playlists')
plt.show()
```

```python
#Line chart

import pandas as pd
import matplotlib.pyplot as plt

plt.plot(data['released_year'], data['artist_count'], marker='o')

# Add title and labels
plt.title('Line Chart from CSV Data')
plt.xlabel('released_year')
plt.ylabel('artist_count')
plt.show()
```
