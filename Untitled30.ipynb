```python
import pandas as pd

# Load the CSV data into a DataFrame
df = pd.read_csv('C:/Users/Rajguru/Downloads/social-media.csv')  # Replace with your actual file name

# Display the original DataFrame (optional)
print("Original DataFrame:")
print(df)

# Specify the condition for rows to delete (e.g., delete rows where Age > 50)
# Modify the condition as needed
rows_to_delete = df[df['Age'] > 40].index

# Drop the specified rows
df = df.drop(rows_to_delete)

# Display the updated DataFrame (optional)
print("\nUpdated DataFrame:")
print(df)

# Save the updated DataFrame back to the CSV file
df.to_csv('user_data.csv', index=False)  # Replace with your actual file name
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df.plot(kind='bar', x='Age', y='TotalLikes', color=['yellow', 'pink'])
plt.title('age wise total likes')
plt.xlabel('Age')
plt.ylabel('TotalLikes')

# Show the chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.hist(df['UsageDuraiton'], bins=5)  
plt.title('UsageDuraiton of conutry')
plt.xlabel('UsageDuraiton')
plt.ylabel('Country')

# Show the histogram
plt.show()
```
