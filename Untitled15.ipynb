```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Sample Data
x = np.random.rand(100)  # Random data for X
y = np.random.rand(100)  # Random data for Y
z = np.random.rand(100)  # Random data for Z

# 3D Scatter Plot
fig = plt.figure(figsize=(12, 8))

# Scatter Plot
ax1 = fig.add_subplot(221, projection='3d')
ax1.scatter(x, y, z, c='r', marker='o')
ax1.set_title('3D Scatter Plot')
ax1.set_xlabel('X-axis')
ax1.set_ylabel('Y-axis')
ax1.set_zlabel('Z-axis')

# 3D Line Graph
ax2 = fig.add_subplot(222, projection='3d')
ax2.plot(x, y, z, color='b')
ax2.set_title('3D Line Graph')
ax2.set_xlabel('X-axis')
ax2.set_ylabel('Y-axis')
ax2.set_zlabel('Z-axis')

# 3D Bar Graph
ax3 = fig.add_subplot(223, projection='3d')
x_bar = [1, 2, 3]
y_bar = [1, 2, 3]
z_bar = [0, 0, 0]
dx = dy = np.array([0.5, 0.5, 0.5])
dz = np.array([1, 2, 3])

ax3.bar3d(x_bar, y_bar, z_bar, dx, dy, dz, color='g', alpha=0.7)
ax3.set_title('3D Bar Graph')
ax3.set_xlabel('X-axis')
ax3.set_ylabel('Y-axis')
ax3.set_zlabel('Z-axis')

# Show the plots
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample Data
subjects = ['Math', 'Science', 'Arts', 'History', 'PE']
students = [25, 30, 15, 20, 18]

# Bar Graph
plt.figure(figsize=(10, 5))
plt.bar(subjects, students, color='skyblue')
plt.title('Number of Students in Each Subject')
plt.xlabel('Subjects')
plt.ylabel('Number of Students')
plt.grid(axis='y')
plt.show()

# Pie Chart
plt.figure(figsize=(8, 8))
plt.pie(students, labels=subjects, autopct='%1.1f%%', startangle=140)
plt.title('Student Distribution by Subject')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()

# Line Graph
# Assuming we have some data for the number of students over 5 years
years = np.array([1, 2, 3, 4, 5])
num_students = np.array([20, 25, 30, 35, 40])

plt.figure(figsize=(10, 5))
plt.plot(years, num_students, marker='o', color='b')
plt.title('Number of Students Over 5 Years')
plt.xlabel('Years')
plt.ylabel('Number of Students')
plt.xticks(years)  # Set x-ticks to years
plt.grid()
plt.show()

# Scatter Plot
# Sample data for hours studied vs. exam scores
hours_studied = np.array([1, 2, 3, 4, 5, 6, 7])
exam_scores = np.array([50, 55, 60, 65, 70, 75, 80])

plt.figure(figsize=(10, 5))
plt.scatter(hours_studied, exam_scores, color='r', s=100)
plt.title('Hours Studied vs. Exam Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Scores')
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Sample data
X = np.array([1, 2, 3, 4, 5, 6, 7])
Y = np.array([50, 55, 60, 65, 70, 75, 80])

# Adding a constant for the intercept
X_with_const = sm.add_constant(X)

# Fit the regression model for Y on X
model_Y_on_X = sm.OLS(Y, X_with_const).fit()

# Fit the regression model for X on Y
X_with_const_Y = sm.add_constant(Y)
model_X_on_Y = sm.OLS(X, X_with_const_Y).fit()

# Generate points for regression lines
X_line = np.linspace(0, 8, 100)
Y_line_Y_on_X = model_Y_on_X.predict(sm.add_constant(X_line))  # Y on X
X_line_X_on_Y = model_X_on_Y.predict(sm.add_constant(Y_line_Y_on_X))  # X on Y

# Plotting
plt.figure(figsize=(12, 6))

# Scatter plot of original data
plt.scatter(X, Y, color='blue', label='Data Points')

# Plotting the regression line of Y on X
plt.plot(X_line, Y_line_Y_on_X, color='red', label='Regression Line of Y on X', linewidth=2)

# Plotting the regression line of X on Y
plt.plot(X_line_X_on_Y, Y_line_Y_on_X, color='green', label='Regression Line of X on Y', linewidth=2)

# Labels and title
plt.title('Regression Lines')
plt.xlabel('Hours Studied (X)')
plt.ylabel('Exam Scores (Y)')
plt.axhline(0, color='black',linewidth=0.5, ls='--')
plt.axvline(0, color='black',linewidth=0.5, ls='--')
plt.grid()
plt.legend()
plt.show()
```
