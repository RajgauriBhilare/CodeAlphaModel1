```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data=pd.read_csv("C:/Users/Rajguru/OneDrive/Desktop/OneDrive/Documents/New folder (2)/User_Data.csv")
data
# Extracting Independent and Dependent Variables  
x = data.iloc[:, [2, 3]].values  # Independent variables (features)
y = data.iloc[:, 4].values       # Dependent variable (target)

# Splitting the dataset into training and test set.
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=0)

# Feature Scaling
from sklearn.preprocessing import StandardScaler
st_x = StandardScaler()

# Apply scaling only to the training set and test set separately to avoid data leakage
x_train = st_x.fit_transform(x_train)  # Fit and transform on training set
x_test = st_x.transform(x_test)        # Only transform on test set (avoid leakage)
print(x_test)
# Fitting K-NN classifier to the training set
from sklearn.neighbors import KNeighborsClassifier
classifier = KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)
classifier.fit(x_train, y_train)  # Train the classifier

# Predicting the test set results
y_pred = classifier.predict(x_test)
print(y_pred)
# Creating the Confusion Matrix
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)

# Displaying the Confusion Matrix
print("Confusion Matrix:")
print(cm)

```
