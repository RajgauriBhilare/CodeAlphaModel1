```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
a = 100  # initial population
b = 0.693  # growth rate (approximation for doubling every hour)
hours = np.arange(0, 10, 0.1)  # time from 0 to 10 hours

# Exponential growth model
population = a * np.exp(b * hours)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(hours, population, label='Population Growth', color='blue')
plt.title('Exponential Growth of Bacteria Over Time')
plt.xlabel('Time (hours)')
plt.ylabel('Population Size')
plt.yscale('log')  # Optional: Use logarithmic scale for better visualization
plt.grid()
plt.legend()
plt.show()
```
