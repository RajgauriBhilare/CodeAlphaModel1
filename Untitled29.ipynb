```python
import numpy as np

# Define the matrix A
A = np.array([[4, 1],
              [2, 3]])

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

# Print the results
print("Eigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
```

```python
import numpy as np

# Function to check if a set of vectors forms a subspace
def is_subspace(vectors):
    # Check for the zero vector
    zero_vector = np.array([0, 0, 0])
    if zero_vector not in vectors:
        return False

    # Check closure under addition
    for i in range(len(vectors)):
        for j in range(len(vectors)):
            sum_vector = vectors[i] + vectors[j]
            if sum_vector[2] != 0:  # Must have z-coordinate as 0
                return False

    # Check closure under scalar multiplication
    for vector in vectors:
        for scalar in [2, -1, 0.5]:  # Example scalars
            scaled_vector = scalar * vector
            if scaled_vector[2] != 0:  # Must have z-coordinate as 0
                return False

    return True

# Define a set of vectors in the XY-plane
vectors = [np.array([1, 2, 0]),
           np.array([3, -1, 0]),
           np.array([-2, 4, 0]),
           np.array([0, 0, 0])]

# Check if the set of vectors forms a subspace
if is_subspace(vectors):
    print("The set of vectors forms a subspace.")
else:
    print("The set of vectors does not form a subspace.")
```
