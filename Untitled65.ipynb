```python
import pyfiglet
from termcolor import colored
import random
import time

# List of names to display
names = ["TUSHAR", "RAHUL", "VIVEK", "AKASH", "RAJ", "SUMIT"]

# List of colors for variety
colors = ['red', 'green', 'yellow', 'blue', 'magenta', 'cyan', 'white']

while True:  # Infinite loop to keep changing names
    for name in names:
        # Generate ASCII art for the current name
        ascii_art = pyfiglet.figlet_format(name)

        # Choose a random color
        colored_ascii = colored(ascii_art, random.choice(colors))

        # Print the colored ASCII art
        print(colored_ascii)

        # Wait for 2 seconds before showing the next name
        time.sleep(2)
```
