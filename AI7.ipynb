```python
import nltk
from nltk.chat.util import Chat, reflections

# Define a list of patterns and responses
chatbot_responses = [
    [
        r"hello|hi|hey",
        ["Hello!", "Hi there!", "Hey!"]
    ],
    [
        r"how are you ?",
        ["I'm good, thanks!", "I'm doing well, how about you?"]
    ],
    [
        r"what is your name ?",
        ["I'm a chatbot.", "I don't have a name, but you can call me ChatGPT."]
    ],
    [
        r"what is your location|your location",
        ["I'm IN THE CPU!", "ON CPU!", "NASHIK!"]
    ],
    [
        r"What's your favorite color?|fav color| favorite color ",
        ["I don't have a favorite color!", "I like all colors, but black is special!", "My favorite color is blue!"]
    ],
    [
        r"bye|goodbye",
        ["Goodbye!", "Bye!", "Take care!"]
    ]
]

# Create chatbot instance
chatbot = Chat(chatbot_responses, reflections)

# Start conversation with the user
print("Hello! I'm a simple chatbot. You can type 'exit' to end the conversation.")
while True:
    user_input = input("Enter Your Message: ")
    
    # Check if user wants to exit
    if user_input.lower() == "exit":
        print("Chatbot: Goodbye!")
        break
    
    # Get the response from the chatbot
    response = chatbot.respond(user_input)
    
    # If no response, the chatbot couldn't understand, so provide a default message
    if not response:
        response = "Sorry, I didn't understand that. Can you please clarify?"
    
    print("Chatbot:", response)
```

```python
import re
import random

# Define a dictionary with patterns and corresponding responses
chatbot_responses = {
    r"hello|hi|hey": ["Hello!", "Hi there!", "Hey!"],
    r"how are you ?": ["I'm good, thanks!", "I'm doing well, how about you?"],
    r"what is your name ?": ["I'm a chatbot.", "I don't have a name, but you can call me ChatGPT."],
    r"what is your location|your location": ["I'm IN THE CPU!", "ON CPU!", "NASHIK!"],
    r"What's your favorite color?|fav color| favorite color ": ["I don't have a favorite color!", "I like all colors, but green is special!", "My favorite color is blue!"],
    r"bye|goodbye": ["Goodbye!", "Bye!", "Take care!"]
}

# Function to match user input to a pattern and return a random response
def get_response(user_input):
    # Convert user input to lowercase for case-insensitive matching
    user_input = user_input.lower()
    
    # Check each pattern in the dictionary
    for pattern, responses in chatbot_responses.items():
        if re.search(pattern, user_input):  # If pattern matches user input
            return random.choice(responses)  # Return a random response from the list
    
    # If no pattern matches, return a default response
    return "Sorry, I didn't understand that. Can you please clarify?"

# Main chatbot loop
def chatbot():
    print("Hello! I'm a simple chatbot. Type 'exit' to end the conversation.")
    
    while True:
        # Get input from the user
        user_input = input("Enter Your Message: ")
        
        # Check if the user wants to exit
        if user_input.lower() == "exit":
            print("Chatbot: Goodbye!")
            break
        
        # Get the chatbot's response based on the user input
        response = get_response(user_input)
        
        # Output the chatbot's response
        print("Chatbot:", response)

# Start the chatbot
chatbot()
```
