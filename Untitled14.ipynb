```python
def calculate_median(numbers):
  numbers.sort()
  mid_index = len(numbers) // 2
  if len(numbers) % 2 == 0:
    return (numbers[mid_index - 1] + numbers[mid_index]) / 2
  else:
    return numbers[mid_index]

numbers = [1, 2, 3, 4, 5]
median = calculate_median(numbers)
print("Median:", median)
```

```python
from collections import Counter

numbers = [1, 2, 3, 2, 4, 5, 2]
frequency_counts = Counter(numbers)
mode = frequency_counts.most_common(1)[0][0]
print("Mode:", mode)
```

```python
def calculate_mode(numbers):
    frequency_counts = {}
    for num in numbers:
        frequency_counts[num] = frequency_counts.get(num, 0) + 1

    max_frequency = max(frequency_counts.values())
    modes = [num for num, frequency in frequency_counts.items() if frequency == max_frequency]
    return modes

numbers = [1, 2, 3, 2, 4, 5, 2]
modes = calculate_mode(numbers)
print("Modes:", modes)

```
