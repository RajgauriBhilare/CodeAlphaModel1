```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# Step 1: Load your dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Employee.csv")

# Step 2: Check for non-numeric columns in features
print("Columns before preprocessing:")
print(data.columns)

# Step 3: Separate features and target
features = data.drop(columns=['LeaveOrNot'])  # Features are everything except 'LeaveOrNot'
target = data['LeaveOrNot']  # 'LeaveOrNot' is the target variable

# Step 4: Convert categorical target column ('LeaveOrNot') to numeric if it's categorical (e.g., 'Yes'/'No')
if target.dtype == 'object':
    label_encoder = LabelEncoder()
    target = label_encoder.fit_transform(target)  # Converts 'Yes'/'No' to 1/0

# Step 5: Handle categorical features using One-Hot Encoding
categorical_columns = features.select_dtypes(include=['object']).columns  # Identify categorical columns

# Perform One-Hot Encoding only for categorical columns
features_encoded = pd.get_dummies(features, columns=categorical_columns, drop_first=True)

# Step 6: Normalize the feature columns (only the features, not the target)
scaler = MinMaxScaler()
normalized_features = scaler.fit_transform(features_encoded)

# Step 7: Convert the normalized features back into a DataFrame
normalized_data = pd.DataFrame(normalized_features, columns=features_encoded.columns)

# Step 8: Add the target column back into the normalized data
normalized_data['LeaveOrNot'] = target

# Step 9: Display the normalized data (with features and target)
print("Normalized data with target:")
print(normalized_data.head())
```

#2.Rescale the features of a dataset to be within the range of [-1, 1]

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Step 1: Load your dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Employee.csv")

# Step 2: Separate features and target (if applicable)
# Assume 'LeaveOrNot' is the target, and we drop it from features
features = data.drop(columns=['LeaveOrNot'])  # Features are everything except 'LeaveOrNot'
target = data['LeaveOrNot']  # Target column

# Step 3: Handle categorical columns
# Identify categorical columns (if any)
categorical_columns = features.select_dtypes(include=['object']).columns

# Apply One-Hot Encoding to categorical columns
features_encoded = pd.get_dummies(features, columns=categorical_columns, drop_first=True)

# Step 4: Rescale the features to the range [-1, 1]
scaler = MinMaxScaler(feature_range=(-1, 1))
normalized_features = scaler.fit_transform(features_encoded)

# Step 5: Convert the normalized features back into a DataFrame
normalized_data = pd.DataFrame(normalized_features, columns=features_encoded.columns)

# Step 6: Add the target column back into the normalized data
normalized_data['LeaveOrNot'] = target

# Step 7: Display the normalized data
print(normalized_data.head())
```

```python
#1 . Standardize a dataset so that each feature has a mean of 0 and a standard deviation of 1.

import pandas as pd
from sklearn.preprocessing import StandardScaler

# Step 1: Load your dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Employee.csv")

# Step 2: Separate features and target (if applicable)
# Assume 'LeaveOrNot' is the target, and we drop it from features
features = data.drop(columns=['LeaveOrNot'])  # Features are everything except 'LeaveOrNot'
target = data['LeaveOrNot']  # Target column

# Step 3: Handle categorical columns
# Identify categorical columns (if any)
categorical_columns = features.select_dtypes(include=['object']).columns

# Apply One-Hot Encoding to categorical columns
features_encoded = pd.get_dummies(features, columns=categorical_columns, drop_first=True)

# Step 4: Standardize the feature columns (only the features, not the target)
scaler = StandardScaler()
standardized_features = scaler.fit_transform(features_encoded)

# Step 5: Convert the standardized features back into a DataFrame
standardized_data = pd.DataFrame(standardized_features, columns=features_encoded.columns)

# Step 6: Add the target column back into the standardized data
standardized_data['LeaveOrNot'] = target

# Step 7: Display the standardized data
print(standardized_data.head())
```
