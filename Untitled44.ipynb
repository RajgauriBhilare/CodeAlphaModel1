```python
import matplotlib.pyplot as plt
data = [7, 8, 5, 9, 6, 4, 12, 15, 8]
plt.boxplot(data)
plt.title("Box Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 35]
sizes = [100, 200, 300, 400, 500]
plt.scatter(x, y, s=sizes, alpha=0.5, color='blue')
plt.title("Bubble Plot")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
sns.heatmap(data, annot=True, cmap='coolwarm')
plt.title("Heat Map")
plt.show()
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt
import numpy as np
data = np.array([[1, 2], [3, 4], [5, 6], [7, 8]])
linkage_data = linkage(data, method='ward')
dendrogram(linkage_data)
plt.title("Dendrogram")
plt.show()
```

```python
from matplotlib_venn import venn3
import matplotlib.pyplot as plt
venn = venn3(subsets=(10, 15, 5, 5, 10, 5, 2), set_labels=('A', 'B', 'C'))
plt.title("Venn Diagram")
plt.show()
```

```python
import squarify
import matplotlib.pyplot as plt
values = [50, 30, 20]
labels = ['A', 'B', 'C']
squarify.plot(sizes=values, label=labels, alpha=0.7)
plt.axis('off')
plt.title("Tree Map")

plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 35]
z = [5, 15, 20, 25, 30]
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, c='blue')
plt.title("3D Scatter Plot")
plt.show()
```
