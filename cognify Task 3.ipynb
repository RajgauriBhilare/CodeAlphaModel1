```python
import numpy as np
import pandas as pd

# Load dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset  (1).csv")
# Mapping Rating text to positive and negative sentiments
positive_ratings = ["Excellent", "Very Good", "Good"]
negative_ratings = ["Average", "Poor", "Very Poor"]

# Count occurrences of each rating category
rating_text_counts = data["Rating text"].value_counts()

# Calculate average length of rating text (as a proxy for review length)
data["Rating_Text_Length"] = data["Rating text"].apply(lambda x: len(str(x).split()))
avg_rating_text_length = data["Rating_Text_Length"].mean()

# Correlation between rating text length and aggregate rating
correlation_length_rating = np.corrcoef(data["Rating_Text_Length"], data["Aggregate rating"])[0, 1]

# Store results
review_analysis = {
    "Rating Text Counts": rating_text_counts.to_dict(),
    "Average Rating Text Length": avg_rating_text_length,
    "Correlation Between Rating Text Length and Rating": correlation_length_rating
}

print(review_analysis)
```

```python
import numpy as np
import pandas as pd

# Load dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset  (1).csv")
# Identify restaurants with highest and lowest number of votes
highest_voted_restaurant = data.loc[data["Votes"].idxmax(), ["Restaurant Name", "Votes", "Aggregate rating"]]
lowest_voted_restaurant = data.loc[data["Votes"].idxmin(), ["Restaurant Name", "Votes", "Aggregate rating"]]

# Correlation between number of votes and aggregate rating
correlation_votes_rating = np.corrcoef(data["Votes"], data["Aggregate rating"])[0, 1]

# Store results
votes_analysis = {
    "Highest Voted Restaurant": highest_voted_restaurant.to_dict(),
    "Lowest Voted Restaurant": lowest_voted_restaurant.to_dict(),
    "Correlation Between Votes and Rating": correlation_votes_rating
}

print(votes_analysis)
```

```python
import numpy as np
import pandas as pd

# Load dataset
data = pd.read_csv("C:/Users/Rajguru/Downloads/Dataset  (1).csv")

# Analyze relationship between price range and online delivery & table booking
price_delivery_table = data.groupby("Price range").agg(
    Online_Delivery_Percentage=("Has Online delivery", lambda x: (x == "Yes").mean() * 100),
    Table_Booking_Percentage=("Has Table booking", lambda x: (x == "Yes").mean() * 100)
).reset_index()

# Correlation between price range and availability of services
correlation_price_delivery = np.corrcoef(data["Price range"], (data["Has Online delivery"] == "Yes"))[0, 1]
correlation_price_table = np.corrcoef(data["Price range"], (data["Has Table booking"] == "Yes"))[0, 1]

# Store results
price_service_analysis = {
    "Price Range vs Online Delivery & Table Booking": price_delivery_table.to_dict(orient="records"),
    "Correlation Between Price Range and Online Delivery": correlation_price_delivery,
    "Correlation Between Price Range and Table Booking": correlation_price_table
}

print(price_service_analysis)
```
