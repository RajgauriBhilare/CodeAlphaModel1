```python
import matplotlib.pyplot as plt
import numpy as np

methods = ['Gradient Boosting', 'Bagging', 'Stacking']
accuracy = [85, 80, 82]
precision = [82, 78, 79]
recall = [88, 82, 84]
f1_score = [85, 80, 81]

x = np.arange(len(methods))  # the label locations
width = 0.2  # the width of the bars

fig, ax = plt.subplots()
bars1 = ax.bar(x - width*1.5, accuracy, width, label='Accuracy')
bars2 = ax.bar(x - width*0.5, precision, width, label='Precision')
bars3 = ax.bar(x + width*0.5, recall, width, label='Recall')
bars4 = ax.bar(x + width*1.5, f1_score, width, label='F1-Score')

ax.set_xlabel('Method')
ax.set_title('Performance Metrics by Method')
ax.set_xticks(x)
ax.set_xticklabels(methods)
ax.legend()

fig.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

methods = ['Gradient Boosting', 'Bagging', 'Stacking']
accuracy = [85, 80, 82]
precision = [82, 78, 79]
recall = [88, 82, 84]
f1_score = [85, 80, 81]

plt.plot(methods, accuracy, marker='o', label='Accuracy')
plt.plot(methods, precision, marker='o', label='Precision')
plt.plot(methods, recall, marker='o', label='Recall')
plt.plot(methods, f1_score, marker='o', label='F1-Score')

plt.xlabel('Method')
plt.ylabel('Score (%)')
plt.title('Performance Comparison of Ensemble Methods')
plt.legend()
plt.grid(True)
plt.show()


```

```python
import matplotlib.pyplot as plt

# Define the data
methods = ['Gradient Boosting', 'Bagging', 'Stacking']
accuracy = [85, 80, 82]
precision = [82, 78, 79]
recall = [88, 82, 84]
f1_score = [85, 80, 81]

# Plot scatter plots
plt.figure(figsize=(10, 6))

plt.scatter(methods, accuracy, color='blue', label='Accuracy', s=100, edgecolor='black')
plt.scatter(methods, precision, color='green', label='Precision', s=100, edgecolor='black')
plt.scatter(methods, recall, color='red', label='Recall', s=100, edgecolor='black')
plt.scatter(methods, f1_score, color='purple', label='F1-Score', s=100, edgecolor='black')

# Add titles and labels
plt.title('Performance Metrics of Ensemble Methods')
plt.xlabel('Method')
plt.ylabel('Score (%)')
plt.legend()
plt.grid(True)
plt.show()
```
