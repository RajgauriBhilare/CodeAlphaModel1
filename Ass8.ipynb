```python
#Use a Kaggle dataset with numerical data (e.g., house prices or ages). How is the distribution of a
#particular variable (e.g., price or age) spread across different intervals? Plot a histogram to visualize this
#distribution.

# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (assuming you've downloaded the Kaggle dataset)
# For demonstration, we're assuming the dataset is named "train.csv"
df = pd.read_csv("C:/Users/Rajguru/Downloads/supermarket_sales - Sheet1.csv")

# Inspect the first few rows to find the column of interest
print(df.head())

```

```python
# Select the 'SalePrice' column (if you're working with house prices)
sale_price = df['Total']

# Plot a histogram to visualize the distribution of SalePrice
plt.figure(figsize=(10, 6))
sns.histplot(sale_price, bins=30, kde=True, color='blue')

# Add title and labels
plt.title('Distribution super market sales', fontsize=16)
plt.xlabel('Unit price', fontsize=14)
plt.ylabel('Quantity', fontsize=14)

# Show the plot
plt.show()
```

```python
#Select a dataset that tracks changes over time (e.g., sales of different product categories over several months).
#How do the sales of each category contribute to the total sales over time? Create a stacked area plot to show this
#trend. Use categorical data 
#overall sales? Create a donut chart to visualize the percentage contribution of each region.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (replace 'train.csv' with your actual dataset path)
df = pd.read_csv("C:/Users/Rajguru/Downloads/cards_data.csv")

# Inspect the first few rows to check the available columns
print(df.head())
```

```python
# Clean the credit_limit column by removing dollar sign and converting to numeric
df['credit_limit'] = df['credit_limit'].replace({'\$': '', ',': ''}, regex=True).astype(float)

# Group by 'card_brand' and calculate total credit limit for each brand
credit_limit_by_brand = df.groupby('card_brand')['credit_limit'].sum()

# Plotting the donut chart
plt.figure(figsize=(8, 8))
plt.pie(credit_limit_by_brand, labels=credit_limit_by_brand.index, autopct='%1.1f%%', startangle=90, pctdistance=0.85)

# Draw the center of the donut by adding a white circle
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
plt.gca().add_artist(centre_circle)

# Add title and show the plot
plt.title('Percentage Contribution of Each Card Brand to Total Credit Limit', fontsize=16)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
