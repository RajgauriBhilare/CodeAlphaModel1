```python
import matplotlib.pyplot as plt

# Data: Heights (in cm) of a sample of 10 students
heights = [150, 155, 160, 165, 170, 175, 180, 185, 190, 195]

# Create histogram
plt.hist(heights, bins=[150, 158, 167, 176, 185, 195], edgecolor='black', alpha=0.7)

# Add titles and labels
plt.title('Histogram of Student Heights')
plt.xlabel('Height Intervals (cm)')
plt.ylabel('Frequency')
plt.xticks([150, 158, 167, 176, 185, 195])

# Show the plot
plt.grid(axis='y')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data: Market share of different companies
companies = ['Company A', 'Company B', 'Company C', 'Company D', 'Company E']
market_share = [25, 35, 20, 15, 5]

# Create a pie chart
plt.figure(figsize=(8, 6))
plt.pie(market_share, labels=companies, autopct='%1.1f%%', startangle=140)

# Add a title
plt.title('Market Share of Different Companies')

# Show the plot
plt.axis('equal')  # Equal aspect ratio ensures the pie chart is circular
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.graphics.mosaicplot import mosaic

# Sample data
data = {
    'Age Group': ['18-24', '18-24', '18-24', '25-34', '25-34', '25-34', '35-44', '35-44', '35-44', '45+'],
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C', 'C'],
    'Count': [30, 20, 50, 40, 30, 30, 20, 40, 40, 70]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Pivot the DataFrame
pivot_df = df.pivot_table(index='Age Group', columns='Product', values='Count', fill_value=0)

# Create the mosaic plot
plt.figure(figsize=(10, 6))
mosaic(pivot_df.stack(), title='Customer Preferences by Age Group and Product')
plt.xlabel('Age Group')
plt.ylabel('Product')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
defect_types = ['Type A', 'Type B', 'Type C', 'Type D', 'Type E']
frequencies = [50, 30, 10, 5, 5]

# Sort data by frequency
sorted_indices = np.argsort(frequencies)[::-1]
sorted_defect_types = [defect_types[i] for i in sorted_indices]
sorted_frequencies = [frequencies[i] for i in sorted_indices]

# Calculate cumulative percentages
cumulative = np.cumsum(sorted_frequencies)
cumulative_percentage = cumulative / cumulative[-1] * 100

# Create the figure and axis
fig, ax1 = plt.subplots()

# Bar plot
ax1.bar(sorted_defect_types, sorted_frequencies, color='b', alpha=0.7)
ax1.set_xlabel('Defect Types')
ax1.set_ylabel('Frequency', color='b')
ax1.tick_params(axis='y', labelcolor='b')

# Create a second y-axis for cumulative percentage
ax2 = ax1.twinx()
ax2.plot(sorted_defect_types, cumulative_percentage, color='r', marker='o', linestyle='--')
ax2.set_ylabel('Cumulative Percentage (%)', color='r')
ax2.tick_params(axis='y', labelcolor='r')

# Add title
plt.title('Pareto Chart of Defect Types')
plt.grid(axis='y')
plt.show()
```
