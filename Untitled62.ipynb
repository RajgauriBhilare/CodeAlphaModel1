```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python
df=pd.DataFrame({'Eduaction':['high school','grduation','postgradutaion','phd']})
df
```

```python
encoder=LabelEncoder()
df['Education_Encoded']=encoder.fit_transform(df['Eduaction'])
print(df)
```

```python
df=pd.DataFrame({'city':['pune','mumbai','delhi']})
df_encoded=pd.get_dummies(df,columns=['city'])
print(df_encoded)
```
