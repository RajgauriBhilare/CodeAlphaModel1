```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
data = {
    "Age": [25, 30, 35, 40, 45],
    "Salary": [40000, 50000, 60000, 70000, 80000],
    "Education": [12, 14, 16, 18, 20],
    "Work_Hours": [50, 48, 45, 40, 35]
}
df = pd.DataFrame(data)

# Compute correlation matrix
correlation_matrix = df.corr()

# Visualize as a heatmap
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
import pandas as pd

# Create example DataFrames
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Carol']})
df2 = pd.DataFrame({'ID': [2, 3, 4], 'Salary': [5000, 6000, 7000]})

# Inner Join
inner_join = pd.merge(df1, df2, on='ID', how='inner')
print(inner_join)
```

```python
# Left Outer Join
left_join = pd.merge(df1, df2, on='ID', how='left')
print("Left Outer Join:\n", left_join)

# Right Outer Join
right_join = pd.merge(df1, df2, on='ID', how='right')
print("\nRight Outer Join:\n", right_join)

# Full Outer Join
full_outer_join = pd.merge(df1, df2, on='ID', how='outer')
print("\nFull Outer Join:\n", full_outer_join)
```
