Implement List Operations(Nested list, Length, Concatenation, Membership ,Iteration ,Indexing and Slicing), List Methods(Add,
Append, Extend & Delete) in python

```python
# 1. Nested List
nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Nested List:", nested_list)
```

```python
# 2. Length of the List
list1 = [1, 2, 3, 4, 5]
length = len(list1)
print("\nLength of list1:", length)
```

```python
# 3. Concatenation of Lists
list2 = [6, 7, 8]
concatenated_list = list1 + list2
print("\nConcatenated List:", concatenated_list)
```

```python
# 4. Membership (Check if an element is in the list)
is_member = 4 in list1  # Check if 3 is in list1
print("\nIs 5 in list1?", is_member)
```

```python
# 5. Iteration (Looping through the list)
print("\nIterating through list1:")
for item in list1:
    print(item, end=" ")
```

```python
# 6. Indexing and Slicing
# Indexing (Accessing elements at a specific index)
first_element = list1[0]  # Accessing first element
last_element = list1[-1]  # Accessing last element
print("\n\nFirst element:", first_element)
print("Last element:", last_element)
```

```python
# Slicing (Getting a sublist from a list)
sliced_list = list1[1:4]  # Get elements from index 1 to 3 (exclusive of index 4)
print("Sliced list (from index 1 to 3):", sliced_list)
```

```python
# 7. List Methods
# Add - Adding an element at a specific position using insert()
list1.insert(2, 10)  # Insert 10 at index 2
print("\nAfter adding 10 at index 2:", list1)
```

```python
# 8.Append - Adding an element to the end of the list
list1.append(11)  # Appending 11 to the end of list1
print("After appending 11:", list1)
```

```python
# 9.Extend - Adding multiple elements at the end of the list
list1.extend([12, 13, 14])  # Extending list1 with another list
print("After extending with [12, 13, 14]:", list1)
```

```python
#10. Delete - Removing an element from the list by value or index
list1.remove(11)  # Removing the element 10
print("After removing 11:", list1)
```

```python
#11. Deleting element by index
del list1[2]  # Deleting the element at index 2
print("After deleting element at index 2:", list1)
```

```python
# 12.Pop - Popping (removing and returning) an element from the end of the list
popped_element = list1.pop()  # Remove and return the last element
print("Popped element:", popped_element)
print("After popping:", list1)
```

```python
# Clear - Clearing all elements from the list
list1.clear()  # Remove all elements
print("After clearing the list:", list1)
```
